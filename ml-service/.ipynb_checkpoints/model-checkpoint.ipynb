```python
data = df.copy()
```

```python
data = data.drop(columns = ['transaction_id', 'ngo_name', 'recipient_id'], errors = 'ignore')
```

```python
data = pd.get_dummies(data, columns = ['recipient_type','region'], dtype = int)
```

```python
data = data.drop(columns=['disbursement_purpose'])

data = pd.get_dummies(data, columns=['payment_mode'], dtype = int)

data['amount_per_beneficiary'] = data['amount'] / data['beneficiary_count']
data = data.drop(columns=['beneficiary_count'], errors = 'ignore')

data['date'] = pd.to_datetime(data['date'])
data['month'] = data['date'].dt.month
data['day_of_week'] = data['date'].dt.dayofweek  
data = data.drop(columns=['date'], errors = 'ignore')
```

```python
labels = data[['is_anomaly', 'anomaly_reason']].copy()
data = data.drop(columns=['is_anomaly', 'anomaly_reason'])
```

```python
data.info()
```

```python
data = pd.get_dummies(data, columns=['ngo_category'], dtype = int)
```

```python
data.info()
```

```python
ngo_ids = data['ngo_id'].copy()
data = data.drop(columns=['ngo_id'])
```
